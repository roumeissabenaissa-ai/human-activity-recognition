```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import random
import numpy as np
import os
import zipfile
```

```python

path = "UCI HAR Dataset/"
```

```python

features = pd.read_csv(path + "features.txt", sep=r"\s+", header=None, names=['Index','FeatureName'])
feature_names = features['FeatureName'].tolist()
```

```python

X_train = pd.read_csv(path + "train/X_train.txt", sep=r"\s+", header=None)
y_train = pd.read_csv(path + "train/y_train.txt", sep=r"\s+", header=None)

X_train.columns = feature_names

```

```python

activity_labels = {
    1: "WALKING",
    2: "WALKING_UPSTAIRS",
    3: "WALKING_DOWNSTAIRS",
    4: "SITTING",
    5: "STANDING",
    6: "LAYING"
}
y_train_named = y_train[0].map(activity_labels)

```

```python

train_data = X_train.copy()
train_data['Activity'] = y_train_named

print(" Dataset chargé avec succès")
print("Train data shape:", train_data.shape)
```

```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split

X = train_data.drop('Activity', axis=1)
y = train_data['Activity']

print("Shape des features :", X.shape)
print("Shape du label :", y.shape)


label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

print("Encodage des activités :")
print(dict(zip(label_encoder.classes_,
               label_encoder.transform(label_encoder.classes_))))


scaler = MinMaxScaler()
X_scaled = pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns
)

print("\nExemple après normalisation :")
print(X_scaled.head())
print("Min global :", X_scaled.min().min())
print("Max global :", X_scaled.max().max())


pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

print("\nAncien nombre de features :", X_scaled.shape[1])
print("Nouveau nombre de features après PCA :", X_pca.shape[1])
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

pca_df = pd.DataFrame(X_pca, columns=['PC1','PC2'])
pca_df['Activity'] = train_data['Activity']

plt.figure(figsize=(10,6))
sns.scatterplot(x='PC1', y='PC2', hue='Activity', data=pca_df, palette='Set1', alpha=0.7)
plt.title("PCA Projection of Training Set")
plt.show()
```

```python

from sklearn.model_selection import StratifiedKFold
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier


from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, roc_curve, auc
)


from sklearn.preprocessing import label_binarize


import numpy as np
import matplotlib.pyplot as plt

import pandas as pd
import time

models = {
    "KNN": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "ANN": MLPClassifier(hidden_layer_sizes=(50,50), max_iter=500, random_state=42)
}


kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

y_bin = label_binarize(y_encoded, classes=np.unique(y_encoded))
n_classes = y_bin.shape[1]


results_list = []

for name, model in models.items():
    print(f"\n--- {name} ---")

    accuracy_list = []
    precision_list = []
    recall_list = []
    f1_list = []
    cm_total = np.zeros((n_classes, n_classes))


    fpr_dict = {i: [] for i in range(n_classes)}
    tpr_dict = {i: [] for i in range(n_classes)}
    roc_auc_dict = {i: [] for i in range(n_classes)}

    start_time = time.time()

    for train_idx, test_idx in kfold.split(X_pca, y_encoded):
        X_train, X_test = X_pca[train_idx], X_pca[test_idx]
        y_train, y_test = y_encoded[train_idx], y_encoded[test_idx]
        y_test_bin = y_bin[test_idx]

        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)

        accuracy_list.append(accuracy_score(y_test, y_pred))
        precision_list.append(precision_score(y_test, y_pred, average='macro', zero_division=0))
        recall_list.append(recall_score(y_test, y_pred, average='macro', zero_division=0))
        f1_list.append(f1_score(y_test, y_pred, average='macro', zero_division=0))
        cm_total += confusion_matrix(y_test, y_pred)


        if hasattr(model, "predict_proba"):
            y_score = model.predict_proba(X_test)
            for i in range(n_classes):
                fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_score[i][:,1] if n_classes==2 else y_score[:,i])
                roc_auc = auc(fpr, tpr)
                fpr_dict[i].append(fpr)
                tpr_dict[i].append(tpr)
                roc_auc_dict[i].append(roc_auc)

    end_time = time.time()
    training_time = end_time - start_time


    accuracy_mean = np.mean(accuracy_list)
    precision_mean = np.mean(precision_list)
    recall_mean = np.mean(recall_list)
    f1_mean = np.mean(f1_list)

    print(f"Accuracy moyenne: {accuracy_mean:.4f}")
    print(f"Precision moyenne: {precision_mean:.4f}")
    print(f"Recall moyenne: {recall_mean:.4f}")
    print(f"F1-score moyen: {f1_mean:.4f}")
    print(f"Matrice de confusion totale:\n{cm_total.astype(int)}")


    results_list.append({
        "Model": name,
        "Accuracy": accuracy_mean,
        "Precision": precision_mean,
        "Recall": recall_mean,
        "F1-score": f1_mean,
        "Training Time (s)": training_time,
        "Confusion Matrix": cm_total.astype(int).tolist()
    })


    if hasattr(model, "predict_proba"):
        plt.figure(figsize=(8,6))
        for i in range(n_classes):
            mean_fpr = np.linspace(0,1,100)
            tprs = []
            for j in range(len(fpr_dict[i])):
                tpr_interp = np.interp(mean_fpr, fpr_dict[i][j], tpr_dict[i][j])
                tpr_interp[0] = 0.0
                tprs.append(tpr_interp)
            mean_tpr = np.mean(tprs, axis=0)
            mean_tpr[-1] = 1.0
            mean_auc = auc(mean_fpr, mean_tpr)
            plt.plot(mean_fpr, mean_tpr, label=f"Classe {i} (AUC = {mean_auc:.2f})")

        plt.plot([0,1],[0,1],'k--')
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.title(f"ROC Curve multi-class: {name}")
        plt.legend()
        plt.show()

df_results = pd.DataFrame(results_list)
df_results.to_csv("model_result.csv", index=False)
print("\nLes résultats ont été enregistrés dans 'model_results.csv'.")
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.metrics import silhouette_score

```

```python


k = 6
kmeans = KMeans(n_clusters=k, random_state=46)
kmeans_labels = kmeans.fit_predict(X_scaled)

silhouette_kmeans = silhouette_score(X_scaled, kmeans_labels)
print(f"Silhouette Score K-Means : {silhouette_kmeans:.3f}")



plt.figure(figsize=(8,5))
sns.scatterplot(x='PC1', y='PC2', hue=kmeans_labels, data=pca_df, palette='Set1', alpha=0.6)
plt.title("K-Means Clustering (2D PCA)")
plt.show()

```

```python

agglo = AgglomerativeClustering(n_clusters=6, linkage='ward')
agglo_labels = agglo.fit_predict(X_scaled)

silhouette_agglo = silhouette_score(X_scaled, agglo_labels)
print(f"Silhouette Score Agglomerative : {silhouette_agglo:.3f}")

pca_df['Cluster'] = agglo_labels
plt.figure(figsize=(8,5))
sns.scatterplot(x='PC1', y='PC2', hue='Cluster', data=pca_df, palette='Set2', alpha=0.6)
plt.title("Agglomerative Clustering (2D PCA)")
plt.show()
```

```python
import scipy.cluster.hierarchy as sch
from scipy.spatial.distance import pdist


distances = pdist(X_scaled, metric='euclidean')


linkage_matrix = sch.linkage(distances, method='ward')


plt.figure(figsize=(12, 6))
sch.dendrogram(linkage_matrix,
               truncate_mode='level',
               p=20,
               leaf_rotation=90,
               leaf_font_size=10)
plt.title("Dendrogram - Agglomerative Clustering")
plt.xlabel("Exemples")
plt.ylabel("Distance")
plt.show()
```

```python

dbscan = DBSCAN(eps=3, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_scaled)


pca_df['DBSCAN_Cluster'] = dbscan_labels

unique_labels = np.unique(dbscan_labels)

n_clusters = len(unique_labels[unique_labels != -1])
n_outliers = list(dbscan_labels).count(-1)

print(f"Nombre de clusters détectés : {n_clusters}")
print(f"Nombre d'outliers : {n_outliers}")


if n_clusters > 1:
    silhouette_dbscan = silhouette_score(X_scaled, dbscan_labels)
    print(f"Silhouette Score DBSCAN : {silhouette_dbscan:.3f}")
else:
    print("Silhouette Score non calculable (moins de 2 clusters)")

plt.figure(figsize=(8,5))
sns.scatterplot(
    x='PC1',
    y='PC2',
    hue='DBSCAN_Cluster',
    data=pca_df,
    alpha=0.7
)

plt.title(f"DBSCAN Clustering (2D PCA)\nClusters = {n_clusters} | Outliers = {n_outliers}")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.legend(title='Cluster', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python

from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.metrics import adjusted_rand_score, adjusted_mutual_info_score


models = {
    "K-Means": kmeans_labels,
    "Agglomerative": agglo_labels,
    "DBSCAN": dbscan_labels
}


results = pd.DataFrame(columns=["Model", "Silhouette", "Davies-Bouldin", "ARI", "AMI"])

for model_name, labels in models.items():


    if len(set(labels)) > 1 and len(set(labels)) != 1 and not all(labels == -1):
        silhouette = silhouette_score(X_scaled, labels)
        davies_bouldin = davies_bouldin_score(X_scaled, labels)
    else:
        silhouette = np.nan
        davies_bouldin = np.nan

    mask = labels != -1
    ari = adjusted_rand_score(y_encoded[mask], labels[mask])
    ami = adjusted_mutual_info_score(y_encoded[mask], labels[mask])


    results = pd.concat([results, pd.DataFrame({
        "Model": [model_name],
        "Silhouette": [silhouette],
        "Davies-Bouldin": [davies_bouldin],
        "ARI": [ari],
        "AMI": [ami]
    })], ignore_index=True)


print("Performances des modèles de clustering :")
print(results)
results.to_csv("clustering_results.csv", index=False)


```
